```python
#2. Check student exam eligibility (attendance + marks)
#Assumed criteria: attendance at least 75% and internal marks at least 40.

attendance = float(input("Enter attendance percentage: "))
marks = int(input("Enter internal marks: "))

if attendance >= 75:
    if marks >= 40:
        print("Eligible for exam")
    else:
        print("Not eligible: Low marks")
else:
    print("Not eligible: Low attendance")
```
