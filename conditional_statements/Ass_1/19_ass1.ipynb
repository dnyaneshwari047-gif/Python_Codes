```python
#19. Write a program to print the day name based on day number (1–7).

num = int(input("Enter number:"))
if num == 1:
    print("Monday")
elif num == 2:
    print("Tuesday")
elif num == 3:
    print("Wednsday")
elif num == 4:
    print("Thursday")
elif num == 5:
    print("Friday")
elif num == 6:
    print("Saturday")
elif num == 7:
    print("Sunday")
else:
    print(" Please enter valid Number!!")
    
```
