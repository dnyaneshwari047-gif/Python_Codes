```python
# 48. Language choice

print("1. English")
print("2. Hindi")
print("3. Marathi")

choice = int(input("Enter your choice (1-3): "))

match choice:
    case 1:
        print("You selected English")
    case 2:
        print("You selected Hindi")
    case 3:
        print("You selected Marathi")
    case _:
        print("Invalid choice!")
```
