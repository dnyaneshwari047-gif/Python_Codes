```python
#43. Write a program to print month days using match case.
choice=int(input("Enter your chocie(1-12):"))
match(choice):
    case 1:print("January!!")
    case 2:print("February!!")
    case 3:print("March!!")
    case 4:print("April!!")
    case 5:print("May!!")
    case 6:print("June!!")
    case 7:print("July!!")
    case 8:print("August!!")
    case 9:print("September!!")
    case 10:print("October!!")
    case 11:print("November!!")
    case 12:print("Decemeber!!")
    case _:print("Please enter valid choice!!")
```
