```python

# 40. Check whether a password is valid using nested if

Email = 'abc@gmail.com'
Password = 'abcc'

email = input("Enter Email: ")
pas = input("Enter Password: ")

if email == Email:
    if pas == Password:
        print("Login Successful!!!")
    else:
        print("Incorrect Password!!")
else:
    print("Incorrect Email!!")
```
