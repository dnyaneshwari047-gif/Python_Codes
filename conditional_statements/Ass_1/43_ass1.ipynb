```python
#42. Write a program to perform calculator operations using match case.
a=float(input("Enter value for a:"))
b=float(input("Enter value for b:"))
print("1.Addition 2.Subtraction 3.Multiplication 4.Division")
choice=int(input("Please enter your choice:"))
match(choice):
    case 1: print("Addition:",a + b)
    case 2:print("Subtraction:",a - b)
    case 3:print("Multiplication:",a*b)
    case 4:print("Division:",a/b)
    case _ :print("Please insert valid choice!!")

```
