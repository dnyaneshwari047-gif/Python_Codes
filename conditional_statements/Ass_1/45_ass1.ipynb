```python
#44. Write a program to check traffic signal actions using match case.
# 44. Write a program to check traffic signal actions using match case.

signal = input("Enter traffic signal color: ").lower()

match signal:
    case "red":
        print("Stop")
    case "yellow":
        print("Wait")
    case "green":
        print("Go")
    case _:
        print("Invalid traffic signal!")
```
