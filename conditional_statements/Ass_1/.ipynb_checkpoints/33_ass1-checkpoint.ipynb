```python
#33. Check whether a person can apply for a loan
#Assumed criteria: age at least 21 and monthly salary at least ₹20,000.

age = int(input("Enter age: "))
salary = int(input("Enter monthly salary: "))

if age >= 21:
    if salary >= 20000:
        print("Eligible to apply for loan")
    else:
        print("Salary is too low")
else:
    print("Age requirement not met")
```
